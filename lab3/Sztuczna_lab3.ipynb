```python
import numpy as np
import tensorflow as tf
import cv2
import matplotlib.pyplot as plt
import tensorflow_datasets as tfds
```

```python
mnist_dataset, mnist_info = tfds.load(name='mnist', with_info=True, as_supervised=True)
mnist_train, mnist_test = mnist_dataset['train'], mnist_dataset['test']
num_validation_samples = 0.1 * mnist_info.splits['train'].num_examples
num_validation_samples = tf.cast(num_validation_samples, tf.int64)
num_test_samples = mnist_info.splits['test'].num_examples
num_test_samples = tf.cast(num_test_samples, tf.int64)
```

```python
def scale(image, label):
  image = tf.cast(image, tf.float32)
  image /= 255
  return image, label

scaled_train_and_validation_data = mnist_train.map(scale)
test_data = mnist_test.map(scale)
```

```python
from tensorflow.python.module.module import valid_identifier
BUFFER_SIZE= 1000
shuffled_train_and_validation_data = scaled_train_and_validation_data.shuffle(BUFFER_SIZE)
validation_data = shuffled_train_and_validation_data.take(num_validation_samples)
train_data = shuffled_train_and_validation_data.skip(num_validation_samples)
```

```python
BATCH_SIZE = 100
train_data = train_data.batch(BATCH_SIZE)
validation_data = validation_data.batch(num_validation_samples)
test_data = test_data.batch(num_test_samples)
validation_inputs, validation_targets = next(iter(validation_data))
print(validation_inputs.shape, validation_targets.shape)
```

```python
input_size = 784
output_size = 10
hidden_layer_size = 50

model = tf.keras.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28, 1)),
  tf.keras.layers.Dense(hidden_layer_size, activation='relu'),
  tf.keras.layers.Dense(hidden_layer_size, activation='relu'),
  tf.keras.layers.Dense(output_size, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
NUM_EPOCHS = 10
early_stopping = tf.keras.callbacks.EarlyStopping(patience=2)
model.fit(train_data,
          epochs=NUM_EPOCHS,
          callbacks=[early_stopping],
          validation_data=(validation_inputs, validation_targets),
          verbose=1
)
```

```python
test_loss, test_accuracy = model.evaluate(test_data)
print('Test loss: {0:.2f}. Test accuracy: {1:.2f}%'.format(test_loss, test_accuracy*100))
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ścieżka do obrazka z cyfrą
img_path = "mnist8.png"

# 1. Wczytanie w skali szarości
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

# sprawdzenie czy obraz się wczytał
if img is None:
    raise ValueError("Nie udało się wczytać obrazka. Sprawdź ścieżkę pliku.")

# 2. Zmiana rozmiaru do 28x28
img_resized = cv2.resize(img, (28, 28))

# 3. Odwrócenie kolorów, jeśli masz czarną cyfrę na białym tle
# MNIST ma zwykle białą cyfrę na czarnym tle
# img_resized = 255 - img_resized

# 4. Normalizacja do zakresu 0-1
img_resized = img_resized.astype("float32") / 255.0

# 5. Dodanie wymiarów wymaganych przez model: (1, 28, 28, 1)
img_input = np.expand_dims(img_resized, axis=0)   # batch
img_input = np.expand_dims(img_input, axis=-1)    # kanał

print("Kształt wejścia do modelu:", img_input.shape)

# 6. Predykcja
predictions = model.predict(img_input)

# 7. Wynik
predicted_digit = np.argmax(predictions)
confidence = np.max(predictions) * 100

print("Przewidziana cyfra:", predicted_digit)
print("Pewność modelu: {:.2f}%".format(confidence))

true_label = 8  # bo np. masz plik mnist8.png

if predicted_digit == true_label:
    print("Model rozpoznał poprawnie ✅")
else:
    print("Model się pomylił ❌")

# 8. Podgląd obrazka
plt.imshow(img_resized, cmap='gray')
plt.title(f"Predykcja: {predicted_digit} ({confidence:.2f}%)")
plt.axis("off")
plt.show()
```
